# Data toolkit


1. What is NumPy, and why is it widely used in Python ?

- NumPy (Numerical Python) is a Python library for numerical computing. It provides powerful multidimensional arrays and functions to perform fast mathematical operations. It is widely used because it offers high performance, efficient memory usage, and supports vectorized operations that speed up computation.

2. How does broadcasting work in NumPy ?

- Broadcasting allows NumPy to perform operations on arrays of different shapes. Smaller arrays are “stretched” automatically to match the shape of the larger array so element-wise operations can occur without manually replicating data.

3.What is a Pandas DataFrame ?

- A DataFrame is a two-dimensional, labeled data structure in Pandas, similar to a table in Excel. It consists of rows and columns, each with a label, and can hold different data types (integers, floats, strings, etc.) in each column.

4. Explain the use of the groupby() method in Pandas ?

- The groupby() method is used to group data based on one or more columns. It helps in performing aggregation operations like sum, mean, count, etc., on each group, enabling powerful data analysis.

5.Why is Seaborn preferred for statistical visualizations ?

- Seaborn is preferred because it provides high-level, aesthetically pleasing statistical plots. It integrates closely with Pandas and can automatically handle complex statistical calculations and themes with minimal code.

6.What are the differences between NumPy arrays and Python lists ?

- NumPy arrays are homogeneous (same data type), while lists can hold mixed data types.

- NumPy arrays consume less memory and are faster for numerical operations.

- NumPy supports vectorized operations, whereas lists require explicit loops.

7.What is a heatmap, and when should it be used ?

- A heatmap is a graphical representation of data where individual values are represented as colors. It is used to visualize correlations, frequency patterns, or intensity across a matrix.

8.What does the term “vectorized operation” mean in NumPy ?

- Vectorized operations allow applying operations on entire arrays without using loops. NumPy performs these operations internally in optimized C code, making them faster and more efficient.

9.How does Matplotlib differ from Plotly ?

- Matplotlib is a static plotting library, suitable for basic and publication-ready plots.
- Plotly, on the other hand, provides interactive, web-based visualizations with zooming, hovering, and animation capabilities.

10. What is the significance of hierarchical indexing in Pandas ?

- Hierarchical indexing (MultiIndex) allows multiple levels of indexing in rows or columns. It enables complex data representation and easier manipulation of high-dimensional data in a 2D structure.

11.What is the role of Seaborn’s pairplot() function ?

- pairplot() visualizes pairwise relationships between numerical variables in a dataset. It creates a grid of scatter plots and histograms, making it useful for exploring data correlations.

12.What is the purpose of the describe() function in Pandas ?

- The describe() function generates summary statistics of numeric columns — such as count, mean, std, min, max, and quartiles — giving a quick overview of the dataset.

13. Why is handling missing data important in Pandas ?

- Missing data can lead to incorrect analysis or model predictions. Pandas provides methods like fillna(), dropna() to handle missing data and maintain data integrity.

14. What are the benefits of using Plotly for data visualization ?

- Plotly allows creating interactive, dynamic, and web-based visualizations. It supports 3D plots, animations, and dashboards, making it ideal for presentation and data exploration.

15.How does NumPy handle multidimensional arrays ?

- NumPy uses the ndarray object to store multidimensional data. It provides efficient indexing, slicing, and broadcasting across dimensions for complex mathematical computations.

16.What is the role of Bokeh in data visualization ?

- Bokeh is a Python library for creating interactive, web-ready visualizations. It integrates with browsers and supports zooming, panning, and streaming data in real time.

17. Explain the difference between apply() and map() in Pandas ?

- map() is used to apply a function to each element in a Series.

- apply() can be used on both Series and DataFrames to apply functions along rows or columns.

18.What are some advanced features of NumPy ?

- Broadcasting

- Vectorization

- Linear algebra operations

- Fourier transforms

- Random number generation

- Integration with C/C++

19.How does Pandas simplify time series analysis ?

-  Pandas provides built-in support for time-series data using DatetimeIndex. It simplifies operations like resampling, shifting, frequency conversion, and rolling statistics.

20.What is the role of a pivot table in Pandas ?

- A pivot table summarizes data by grouping and aggregating values, similar to Excel pivot tables. It helps analyze relationships and patterns efficiently.

21.Why is NumPy’s array slicing faster than Python’s list slicing ?

- NumPy arrays are stored in contiguous memory blocks and use optimized C code for slicing, while Python lists are stored as objects, making slicing slower.

22. What are some common use cases for Seaborn ?

- Visualizing correlations (heatmaps, pairplots)

- Analyzing distributions (histplots, boxplots, violin plots)

- Comparing categories (bar plots, swarm plots)

- Regression and statistical relationships

```python
'''1.How do you create a 2D NumPy array and calculate the sum of each row. '''

import numpy as np

# Create a 2D NumPy array
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

# Calculate the sum of each row
row_sum = np.sum(arr, axis=1)

# Display the array and result
print("2D Array:")
print(arr)
print("\nSum of each row:")
print(row_sum)
```

```python
'''2.Write a Pandas script to find the mean of a specific column in a DataFrame .'''

import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Amit', 'Ravi', 'Sita', 'Rina', 'Karan'],
    'Marks': [85, 90, 78, 88, 95],
    'Age': [20, 21, 19, 22, 20]
}

df = pd.DataFrame(data)

# Calculate the mean of the 'Marks' column
mean_marks = df['Marks'].mean()

# Display the DataFrame and the result
print("DataFrame:")
print(df)
print("\nMean of 'Marks' column:", mean_marks)
```

```python
'''3.Create a scatter plot using Matplotlib.'''
import matplotlib.pyplot as plt

# Sample data
x = [10, 20, 30, 40, 50]
y = [15, 25, 20, 35, 45]

# Create scatter plot
plt.scatter(x, y, color='blue', marker='o')

# Add labels and title
plt.title("Scatter Plot Example")
plt.xlabel("X Values")
plt.ylabel("Y Values")

# Display the plot
plt.show()
```

```python
'''4.How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap '''

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create a sample DataFrame
data = {
    'Maths': [78, 85, 96, 80, 86],
    'Science': [84, 95, 89, 83, 90],
    'English': [70, 88, 82, 75, 85]
}

df = pd.DataFrame(data)

# Calculate the correlation matrix
corr_matrix = df.corr()

# Display correlation matrix
print("Correlation Matrix:")
print(corr_matrix)

# Visualize with Seaborn heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')

# Add title
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
'''5.Generate a bar plot using Plotly ?'''

import plotly.express as px
import pandas as pd

# Create sample data
data = {
    'Subject': ['Maths', 'Science', 'English', 'History', 'Computer'],
    'Marks': [85, 90, 78, 88, 95]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Create a bar plot using Plotly
fig = px.bar(df, x='Subject', y='Marks', color='Subject',
             title="Marks Obtained in Each Subject")

# Display the bar plot
fig.show()
```

```python
'''6.Create a DataFrame and add a new column based on an existing column.'''

import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Amit', 'Ravi', 'Sita', 'Rina', 'Karan'],
    'Marks': [80, 90, 75, 85, 95]
}

df = pd.DataFrame(data)

# Add a new column based on an existing column
# Let's add a 10% bonus to each student's marks
df['Final_Marks'] = df['Marks'] * 1.10

# Display the DataFrame
print("Updated DataFrame:")
print(df)
```

```python
'''7.Write a program to perform element-wise multiplication of two NumPy arrays.'''
import numpy as np

# Create two NumPy arrays
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])

# Element-wise multiplication
result = arr1 * arr2   # or np.multiply(arr1, arr2)

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Element-wise Multiplication:", result)

```

```python
'''8.Create a line plot with multiple lines using Matplotlib.'''

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]
y3 = [3, 5, 2, 6, 9]

# Create the plot
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')
plt.plot(x, y3, label='Line 3')

# Add title and labels
plt.title("Multiple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Show legend
plt.legend()

# Display the plot
plt.show()
```

```python
'''9.Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold '''


import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 32, 18, 47],
    'Score': [85, 90, 70, 95]
}

df = pd.DataFrame(data)

# Define a threshold
threshold = 30

# Filter rows where 'Age' is greater than the threshold
filtered_df = df[df['Age'] > threshold]

print("Original DataFrame:")
print(df)
print("\nFiltered DataFrame (Age > 30):")
print(filtered_df)
```

```python
'''10.Create a histogram using Seaborn to visualize a distribution.'''

import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = [12, 15, 14, 10, 12, 15, 18, 19, 10, 12, 16, 14, 15, 18, 20]

# Create histogram
sns.histplot(data, bins=5, kde=True, color='skyblue')

# Add title and labels
plt.title("Histogram of Sample Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

# Display the plot
plt.show()
```

```python
'''11. Perform matrix multiplication using NumPy ?'''

import numpy as np

# Define two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Method 1: Using np.dot()
result_dot = np.dot(A, B)

# Method 2: Using the @ operator
result_at = A @ B

# Method 3: Using np.matmul()
result_matmul = np.matmul(A, B)

print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)
print("\nMatrix Multiplication Result (np.dot):")
print(result_dot)
```

```python
'''13.Create a 3D scatter plot using Plotly.'''

import plotly.express as px
import pandas as pd

# Sample data
data = {
    'X': [1, 2, 3, 4, 5, 6],
    'Y': [10, 11, 12, 13, 14, 15],
    'Z': [5, 6, 7, 8, 9, 10],
    'Category': ['A', 'B', 'A', 'B', 'A', 'B']
}

df = pd.DataFrame(data)

# Create 3D scatter plot
fig = px.scatter_3d(df, x='X', y='Y', z='Z', color='Category', size='Z',
                    symbol='Category', title='3D Scatter Plot Example')

# Show plot
fig.show()
```
